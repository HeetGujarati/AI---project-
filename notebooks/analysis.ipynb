# LoRA+ Empirical Analysis & Results Notebook
This notebook analyzes the empirical performance of **LoRA+** against standard LoRA on **NVIDIA GeForce RTX 4060 (8GB)** using `Qwen/Qwen2.5-1.5B`.

```python
import json
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load comparison metrics
metrics_path = os.path.join("..", "experiments", "results", "comparison_metrics.json")
if os.path.exists(metrics_path):
    with open(metrics_path, "r") as f:
        data = json.load(f)
    df = pd.DataFrame(data)
    display(df[["experiment_name", "method", "lr_ratio", "train_loss", "eval_loss", "eval_perplexity", "total_train_time_sec", "peak_memory_allocated_mb"]])
else:
    print("Metrics file not found. Run scripts/run_experiments.py first.")
```

```python
# Plot validation loss across ratios
if os.path.exists(metrics_path):
    plt.figure(figsize=(8, 4.5))
    sns.barplot(data=df, x="experiment_name", y="eval_loss", palette="Blues_d")
    plt.title("Validation Loss Across LoRA and LoRA+ Configurations")
    plt.ylabel("Validation Loss")
    plt.show()
```
